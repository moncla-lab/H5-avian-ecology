```python
def fasta_to_df(path_to_fasta):
    FASTANAME = path_to_fasta

    fasta_data = []
        
    with open(FASTANAME) as f:
        header = ""
        sequence = ""
        for line in f:
            if line.startswith(">"):
                if header != "":
                    fasta_data.append({"header": header, "sequence": sequence})
                header = line.strip() 
                sequence = ""
            else:
                sequence += line.strip()
        fasta_data.append({"header": header, "sequence": sequence}) #last line
    
    fasta_df = pd.DataFrame(fasta_data)

    return(fasta_df)
```

```python
# Function: Make column for accession number, Genome ID, and strain

def fasta_identifiers(fasta_df):

    # Make a copy of fasta_df
    fasta_df_updt = fasta_df.copy()
    
    # Use regex to create a match column for each accession type in fasta_df
    fasta_df_updt['genbank_match'] = fasta_df_updt['header'].str.extract(r"(\w[A-Z]\d{6})")
    fasta_df_updt['bvbrc_match'] = fasta_df_updt['header'].str.extract(r"(11320.\d{6})")
    fasta_df_updt['gisaid_match'] = fasta_df_updt['header'].str.extract(r"(a\/\w+\/\w+\/\w+\/\d{4}\|\d{4}-\d{2}-\w{2}\|\w+\|\w+\|\w+)")
    fasta_df_updt['gisaid_match'] = fasta_df_updt['gisaid_match'].str.replace("original", "", regex = False)
    fasta_df_updt['gisaid_match'] = fasta_df_updt['gisaid_match'].str.replace("usda", "22", regex = False)

    fasta_df_updt["sequence"] = fasta_df_updt["sequence"].str.upper()

    return(fasta_df_updt)
```

```python
# Attempt no. 3 by Clara

def filtering_fasta_sequences(fasta_df, metadata_df):

    # Copy metadata_df to avoid modifying the original
    merged_fasta_metadata = metadata_df.copy()

    # Create a column in metadata_df to store sequences
    merged_fasta_metadata['sequence'] = None

    #Iterate over metadata_df and match with fasta_df
    for idx, metadata_row in metadata_df.iterrows():
        # Try matching each type of accession from metadata_df to fasta_df
        metadata_genbank_acc = metadata_row["genbank_acc"]
        metadata_bvbrc_acc = metadata_row['Genome ID']
        metadata_gisaid_strain = metadata_row['strain']
        
        # Look for the GenBank accession
        if pd.notna(metadata_genbank_acc):
            matching_rows = fasta_df[fasta_df['genbank_match'] == metadata_genbank_acc]
            #print(len(matching_rows), matching_rows)
            if not matching_rows.empty and len(matching_rows) == 1:
                merged_fasta_metadata.at[idx, 'sequence'] = matching_rows['sequence'].iloc[0]
            elif not matching_rows.empty and len(matching_rows) > 1:
                genbank_row = matching_rows[matching_rows["bvbrc_match"].isna()]
                #print(genbank_row)
                #print(genbank_row["sequence"])
                merged_fasta_metadata.at[idx, 'sequence'] = genbank_row['sequence'].iloc[0]
        
        # Look for the BV-BRC accession
        if pd.notna(metadata_bvbrc_acc):
            metadata_bvbrc_acc = str(metadata_bvbrc_acc)
            matching_rows = fasta_df[fasta_df['bvbrc_match'] == metadata_bvbrc_acc]
            if not matching_rows.empty:
                #print(matching_rows)
                #print(matching_rows["sequence"])
                merged_fasta_metadata.at[idx, 'sequence'] = matching_rows['sequence'].iloc[0]
        
        # Look for the GISAID strain
        if pd.notna(metadata_gisaid_strain):
            matching_rows = fasta_df[fasta_df['gisaid_match'] == metadata_gisaid_strain]
            if not matching_rows.empty:
                #print(matching_rows)
                #print(matching_rows["sequence"])
                merged_fasta_metadata.at[idx, 'sequence'] = matching_rows['sequence'].iloc[0]

        else:
            print(f"no matching sequence for {idx}")
    
    return merged_fasta_metadata
```

```python
# Function to write a new fasta with just the sequences in the metadata
# Adapted from Maria

def fasta_writer(path, filename, df):
            
    try:  
        os.mkdir(path)

    except OSError as error:
        pass

    with open(f"{path}{filename}", "w") as f:
        for index, row in df.iterrows():
            f.write(f"{row['header']}\n")
            f.write(f"{row['sequence']}\n")
```
