```python
# Function for making a dictionary of new hosts not already in the complete list
def new_hosts(addition_df, base_df):
    new_series = []

    for i in addition_df['0']:
        i = str(i).strip()
        if i not in set(base_df['0'].str.strip()):
            new_series.append(i)
        else:
            pass

    new_df = pd.DataFrame(new_series, columns=['0'])

    #print(new_df)

    combined_df = pd.concat([base_df, new_df], axis = 0, ignore_index = True)

    return(combined_df)
```
